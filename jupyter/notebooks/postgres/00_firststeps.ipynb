<h5>Há duas formas de inserir valores em tabelas, explicito e implicito</h5>

```bash
INSERT INTO weather VALUES ('San Francisco', 46, 50, 0.25,
'1994-11-27'); #implicito, aqui você tem que decorar a ordem das colunas

INSERT INTO weather (city, temp_lo, temp_hi, prcp, date)
VALUES ('San Francisco', 43, 57, 0.0, '1994-11-29'); #explicito, associa a ordem da colunas a ordem dos valores.

INSERT INTO weather (date, city, temp_hi, temp_lo)
VALUES ('1994-11-29', 'Hayward', 54, 37); #e pode omitir algumas colunas se precisar.
```

<h5>Para você não precisar ficar fazendo a mesma query toda vez, recicle, crie uma <b>view</b></h5>

```bash
CREATE VIEW myview AS
SELECT name, temp_lo, temp_hi, prcp, date, location
FROM weather, cities
WHERE city = name;

SELECT * FROM myview;
```

<h5>Nas camadas de aplicação é nescessários fazer validações, alguns ORM contam com essa funçaõ ou voocê mesmo pode implentar, porém não é dever do postgres. podemos criar essas validações na transação para evitar o ROLLBACK total</h5>

<h5>SAVEPOINT svp</h5>
Tudo antes do save point é preservado.
<h5>ROLLBACK TO svp</h5>
Tudo entre o do rollback to e savepoint é descardato.

```bash
BEGIN;
UPDATE Alice -100;    -- preservado
SAVEPOINT aqui;
UPDATE Bob +100;      -- descartado
ROLLBACK TO aqui;
UPDATE Wally +100;    -- preservado
COMMIT;
```
