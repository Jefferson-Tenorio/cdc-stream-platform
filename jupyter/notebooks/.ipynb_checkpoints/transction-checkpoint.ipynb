# Transações no PostgreSQL

## O que é uma Transação?

Uma transação é o **mecanismo do banco** que agrupa múltiplas operações em um único bloco de execução.

A **unidade lógica de trabalho** é o conceito — operações que fazem sentido juntas:

```sql
-- Transferência bancária: as duas operações precisam acontecer juntas
UPDATE conta SET saldo = saldo - 100 WHERE id = 1;  -- debita
UPDATE conta SET saldo = saldo + 100 WHERE id = 2;  -- credita
```

A **transação** é o mecanismo que agrupa essa lógica:

```sql
BEGIN;
    UPDATE conta SET saldo = saldo - 100 WHERE id = 1;
    UPDATE conta SET saldo = saldo + 100 WHERE id = 2;
COMMIT;  -- ou ROLLBACK para desfazer
```

### Relação entre os conceitos

```
Você define   →  a unidade lógica de trabalho (o que agrupa)
Você usa      →  a transação para agrupar (BEGIN/COMMIT)
Banco garante →  ACID sobre essa transação
```

---

## Propriedades ACID

ACID são as garantias que o banco entrega automaticamente sobre qualquer transação — não é algo que você escolhe, é uma propriedade da transação.

- **Atomicity** — tudo ou nada. Se uma das operações falha, todas as anteriores são revertidas. Não existe estado intermediário persistido.

- **Consistency** — o banco sai de um estado válido e chega em outro estado válido. Constraints, foreign keys e regras de negócio definem o que é "válido".

- **Isolation** — transações concorrentes não se enxergam mutuamente até commitarem. Implementado via MVCC no PostgreSQL.

- **Durability** — depois do commit, os dados sobrevivem a crash. Garantido pelo WAL (Write-Ahead Log).

### Níveis de Isolamento

O "I" do ACID não é binário — cada nível abre a porta para anomalias diferentes:

| Nível | Dirty Read | Non-repeatable Read | Phantom Read |
|---|---|---|---|
| READ COMMITTED | ✗ | possível | possível |
| REPEATABLE READ | ✗ | ✗ | possível |
| SERIALIZABLE | ✗ | ✗ | ✗ |

> O PostgreSQL usa `READ COMMITTED` por padrão.

## Autocommit vs Transação Explícita

### Autocommit (padrão do PostgreSQL)

Cada statement é automaticamente uma transação própria:

```sql
UPDATE conta SET saldo = saldo - 100 WHERE id = 1;
-- commit automático aqui
UPDATE conta SET saldo = saldo + 100 WHERE id = 2;
-- commit automático aqui
```

**Problema:** se o servidor cair entre os dois statements, o dinheiro some — cada operação commitou independentemente.

### Transação Explícita

```sql
BEGIN;
    UPDATE conta SET saldo = saldo - 100 WHERE id = 1;
    UPDATE conta SET saldo = saldo + 100 WHERE id = 2;
COMMIT;
```

Se o servidor cair antes do `COMMIT`, o banco reverte tudo automaticamente.

### Quando usar cada um

| | Autocommit | Explícita |
|---|---|---|
| BEGIN/COMMIT | implícito | você escreve |
| Atomicidade entre statements | não | sim |
| Quando usar | operações independentes | operações interdependentes |

**Regra prática:** se a operação X falhar, a operação Y ainda faz sentido ter acontecido? Se **não** — use transação explícita.

### Contextos comuns

**Autocommit faz sentido para:**
- SELECTs — leitura pura
- Operações independentes entre si
- Scripts de carga onde cada linha é independente
- DDL (`CREATE TABLE`, `ALTER TABLE`)

**Transação explícita faz sentido para:**
- Operações interdependentes entre tabelas
- Quando precisa de rollback em caso de erro parcial
- Aplicações web — ORMs como SQLAlchemy abrem transação por request por padrão
